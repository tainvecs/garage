## Load Packages and Set Arguments

```python
import json
import re
import uuid
```

```python
processed_data_path = "processed-news_category-100.example.json"
processed_data_es_path = "processed-news_category-100.es.example.json"
```

## Processed Data
- The test data is forked from News Category Dataset.
- It contains around 200k news headlines from the year 2012 to 2018 obtained from HuffPost.
- The data is processed with another notebook.

```python
# load data
processed_data_list = []
with open(processed_data_path, 'r') as in_file: 
    processed_data_list = json.load(in_file)

# print example
print(json.dumps(processed_data_list[0], indent=4))
```

## Processed Data for Elasticsearch Indexing

```python
processed_data_es_list = []

for data in processed_data_list: 
    processed_data_es_list.append(
        {
            "index":{
                "_id":data['uuid']
            }
        }
    )
    processed_data_es_list.append(data)
```

```python
# json data -> str
out_jstr_list = []
for d in processed_data_es_list: 
    out_jstr_list.append(json.dumps(d))

# joined and format output str
joined_out_str = "{body}\n".format(
    body='\n'.join(out_jstr_list)
)

# write output str
with open(processed_data_es_path, 'w') as ofile: 
    ofile.write(joined_out_str)
```
